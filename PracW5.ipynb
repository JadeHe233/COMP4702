Q1

```python
import pandas as pd
import numpy as np

df = pd.read_csv("w3classif.csv")

# Extract labels
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

from sklearn.model_selection import train_test_split


splits = []

for i in range(10):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, shuffle=True
    )
    
    splits.append((X_train, X_test, y_train, y_test))
```

Q2

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

results = []

for X_train, X_test, y_train, y_test in splits:
    
    knn = KNeighborsClassifier(n_neighbors=3)
    knn.fit(X_train, y_train)

    y_pred = knn.predict(X_test)
    
    train_accuracy = accuracy_score(y_train, knn.predict(X_train))
    test_accuracy = accuracy_score(y_test, y_pred)

    train_error = 1 - train_accuracy
    test_error = 1 - test_accuracy
    results.append((train_error, test_error))

print("Train errors:", [f"{r[0]*100:.2f}%" for r in results])
print("Test errors:", [f"{r[1]*100:.2f}%" for r in results])

print(f"Mean train error: {np.mean([r[0] for r in results])*100:.2f}%")
print(f"Mean test error: {np.mean([r[1] for r in results])*100:.2f}%")
```

Train error increased, but test error decreased.

Q3

```python
splits_2 = []

for i in range(10):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.1, shuffle=True
    )
    
    splits_2.append((X_train, X_test, y_train, y_test))

results_2 = []

for X_train, X_test, y_train, y_test in splits_2:
    
    knn = KNeighborsClassifier(n_neighbors=3)
    knn.fit(X_train, y_train)

    y_pred = knn.predict(X_test)
    
    train_accuracy = accuracy_score(y_train, knn.predict(X_train))
    test_accuracy = accuracy_score(y_test, y_pred)

    train_error = 1 - train_accuracy
    test_error = 1 - test_accuracy
    results_2.append((train_error, test_error))

print("Train errors:", [f"{r[0]*100:.2f}%" for r in results_2])
print("Test errors:", [f"{r[1]*100:.2f}%" for r in results_2])

print(f"Mean train error: {np.mean([r[0] for r in results_2])*100:.2f}%")
print(f"Mean test error: {np.mean([r[1] for r in results_2])*100:.2f}%")
```

Both errors slightly increased.

Q4

```python
train_std_1 = np.std([r[0] for r in results])
test_std_1 = np.std([r[1] for r in results])

train_std_2 = np.std([r[0] for r in results_2])
test_std_2 = np.std([r[1] for r in results_2])

print(f"Train error std (30% test): {train_std_1*100:.2f}%")
print(f"Test error std (30% test): {test_std_1*100:.2f}%")
print(f"Train error std (10% test): {train_std_2*100:.2f}%")
print(f"Test error std (10% test): {test_std_2*100:.2f}%")
```

There are more fluctuations in test error for test size = 10%. 

Q5

```python
from sklearn.model_selection import cross_val_score

knn = KNeighborsClassifier(n_neighbors=3)

cv_scores = cross_val_score(knn, X, y, cv=10)
errors = 1 - cv_scores

print("Errors:", errors)
print(f"Mean error: {np.mean(errors)*100:.2f}%")
print(f"Std error: {np.std(errors, ddof=1)*100:.2f}%")
```
